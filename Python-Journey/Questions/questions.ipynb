# Question 2

```python
import re
def common_words(row):
    sentence1 = row['sentence_1']
    sentence2 = row['sentence_2']

    pattern = '\w+'
    lst1 = re.findall(pattern , sentence1)
    lst2 = re.findall(pattern , sentence2)
    st1 = set(lst1)
    st2 = set(lst2)

    result = st1.intersection(st2)

    final_result = {}

    for i in lst2:
        if(i in result):
            if(i in final_result):
                final_result[i]+=1
            else:
                final_result[i]=1

    final_result = dict(sorted(final_result.items()))
    return final_result

df2 = pd.read_csv('/Users/pranavaditya/Desktop/Data-Engineering-Journey/question_02/data.csv')
df2['common_words']=df2.apply(common_words,axis=1)
df2
```

```python
import re 

def is_palindrome(check):
    check = check.lower()
    lst = re.findall(r'\w+',check)
    lst = ''.join(lst)
    lst = lst.replace(" ","")

    return lst[::] == lst[::-1]


df3 = pd.read_csv('/Users/pranavaditya/Desktop/Data-Engineering-Journey/question_03/data.csv')
df3['is_palindrome']=df3['text'].apply(is_palindrome)
df3
```

# Question 4 

```python
df4 = pd.read_csv('/Users/pranavaditya/Desktop/Data-Engineering-Journey/question_04/data.csv')
df4

from collections import defaultdict

def freq_chars(df4):
    words = df4['word']
    result = []
    freq = {}
    for word in words:
        for char in word:
            if char in freq:
                freq[char] +=1
            else:
                freq[char] = 1
        result.append({word:freq})
        freq={}
    return result

def is_anagram(lst):

    lst = freq_chars(lst)
    result = defaultdict(list)

    for i in range(0,len(lst)-1):
        dict1_keys  = list(lst[i].keys())
        dict1_keys = dict1_keys[0]

        if not any(dict1_keys in values for values in result.values()):
            result[i].append(dict1_keys)
            for j in range(i+1,len(lst)):
                dict2_keys =  list(lst[j].keys())
                dict2_keys = dict2_keys[0]

                if(len(dict1_keys)!=len(dict2_keys)):
                    continue

                dict1_values = lst[i][dict1_keys]
                dict2_values = lst[j][dict2_keys]
                flag = True

                for val in dict1_values:
                    if(dict1_values.get(val)!=dict2_values.get(val)):
                        flag = False
                        break
                if(flag == True):
                    result[i].append(dict2_keys)
    return dict(result)



final = is_anagram(df4)
df = pd.DataFrame(final.items(),columns = ['group_id','words'])
df['group_length']=len(df['words'])
df
```

```python
df5 = pd.read_csv('/Users/pranavaditya/Desktop/Data-Engineering-Journey/question_05/data.csv')
df5


def reverse_word(text):
    lst = list(text)
    i = 0
    j = len(lst)-1
    while(i<j):
        temp = lst[i]
        lst[i] = lst[j]
        lst[j] = temp
        i+=1
        j-=1
    lst = ''.join(lst)
    return lst
    
def reverse_string(lst):
    lst = lst.split(' ')
    result = []
    for i in lst:
        result.append(reverse_word(i))
    return ' '.join(result)

final = reverse_string('Streamlit apps are interactive')

df5['reversed_string']=df5['sentence'].apply(reverse_string)
df5
```

# Question 6

```python
from datetime import datetime as dt
import json
import re

df6=pd.read_csv('/Users/pranavaditya/Desktop/Data-Engineering-Journey/question_06/data.csv')
df6

def remove_words_with_digits(text):
      return re.findall(r"\b[A-Za-z]+\b",text)
      
def bucket(time):
    if(time>=dt.strptime('09:00:00', "%H:%M:%S").time() and dt.strptime('10:00:00', "%H:%M:%S").time()>time):
        return '09:00 - 10:00'
    elif(time>=dt.strptime('10:00:00', "%H:%M:%S").time() and dt.strptime('11:00:00', "%H:%M:%S").time()>time):
        return '10:00 - 11:00'
    elif(time>=dt.strptime('11:00:00', "%H:%M:%S").time() and dt.strptime('12:00:00', "%H:%M:%S").time()>time):
        return '11:00 - 12:00'
    elif(time>=dt.strptime('12:00:00', "%H:%M:%S").time() and dt.strptime('13:00:00', "%H:%M:%S").time()>time):
            return '12:00 - 13:00'
    elif(time>=dt.strptime('13:00:00', "%H:%M:%S").time() and dt.strptime('14:00:00', "%H:%M:%S").time()>time):
            return '13:00 - 14:00'
    elif(time>=dt.strptime('14:00:00', "%H:%M:%S").time() and dt.strptime('15:00:00', "%H:%M:%S").time()>time):
            return '14:00 - 15:00'
    elif(time>=dt.strptime('15:00:00', "%H:%M:%S").time() and dt.strptime('16:00:00', "%H:%M:%S").time()>time):
                return '15:00 - 16:00'
        
    
def output(js):
    dict = json.loads(js)
    time  = dt.fromisoformat(dict['timestamp']).time()
    text = dict['text']
    return time , text

def freq_words(lst):
      freq={}
      for i in lst:
            if(i.lower() in freq):
                  freq[i]+=1
            else:
                  freq[i] = 1
      return freq

def sortIt(dic):
      return dict(sorted(dic.items(),key=lambda x :(-x[1],x[0]))[:1])

df6[['time','text']]=df6['json_data'].apply(output).apply(pd.Series)
df6['bucket'] = df6['time'].apply(bucket)

df6= df6.groupby('bucket')['text'].agg(' '.join)
df6=df6.apply(remove_words_with_digits)

df6 = df6.apply(freq_words)
df6.apply(sortIt)
```

# Question 7

```python
import pandas as pd
import numpy as np

df7 = pd.read_csv('/Users/pranavaditya/Desktop/Data-Engineering-Journey/question_07/data.csv')
temp = df7
df7 = df7.rename(columns = {'holding_cost':'holding_rate'})

df7['holding_cost']=df7['holding_rate']*df7['unit_cost']
df7 = df7.groupby('sku').agg({
    'demand':'sum',
    'holding_cost':'mean',
    'order_cost':'mean'
})

df7['demand']=df7['demand']*12
df7['EOQ']=np.sqrt((2*df7['demand']*df7['order_cost'])/df7['holding_cost'])
df7


def simulation(row):

    inventory = row['EOQ']
    demand=row['demand_list']

    stockout = 0
    i = 0
    while(i<180):
        inventory-=demand[i%len(demand)]
        if(inventory<=0):
            lead_time = np.random.randint(2,8)
            remaining = 179-i
            effective  = min(lead_time,remaining)
            stockout+=effective+1

            for j in range(1,effective+1):
                inventory-=demand[(i+j)%len(demand)]
                
            inventory+=row['EOQ']

            i+=effective+1
        else:
            i+=1

    return stockout

temp = temp.groupby('sku')['demand'].agg(list)
df7['demand_list']=temp
final= df7[['EOQ','demand_list']]
final

s = df7.apply(simulation,axis=1)
final['stockout_days']=s
result =  final[['EOQ','stockout_days']]
result
```
