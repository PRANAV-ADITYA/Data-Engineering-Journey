```python
from pdf2image import convert_from_path
from PIL import Image
import numpy as np
import cv2
import pytesseract as pc

images = convert_from_path('/Users/pranavaditya/Desktop/Data-Engineering-Journey/Projects/Medical Prescription Project/Backend/pd_1.pdf')


def preprocess_image(img):
    processed_image=[]
    for im in img:
        gray=cv2.cvtColor(np.array(im),cv2.COLOR_BGR2GRAY)
        resized = cv2.resize(gray,None,fx=1.5,fy=1.5,interpolation=cv2.INTER_LINEAR)
        processed_image.append(cv2.adaptiveThreshold(resized,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,61,11))
    return processed_image

def extract_text(lst):
    big_text = ''
    for image in lst:
        text = pc.image_to_string(image,lang = 'eng')
        big_text+=text+'\n'
    return big_text

img = preprocess_image(images)
print(extract_text(img))
text = extract_text(img)
```
